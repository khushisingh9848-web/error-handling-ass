Ques1->: What is the difference between multithreading and multiprocessing?

--->Difference between Multithreading and Multiprocessing:

Multithreading:
In multithreading, a single process (program) is divided into multiple smaller threads that run at the same time. These threads share the same memory and resources of the main process.
Example: Different threads in a web browser — one for loading a page, one for playing a video, and one for responding to clicks.

Multiprocessing:
In multiprocessing, multiple processes run at the same time, and each process has its own separate memory and resources. This allows true parallel execution on multiple CPU cores.
Example: Running multiple programs like a music player, a web browser, and a game at the same time.



Ques2->: What are the challenges associated with memory management in Python?

-->Challenges associated with memory management in Python:

Garbage collection overhead:
Python automatically removes unused objects, but sometimes this process can slow down the program.

Memory leaks:
If references to unused objects are not properly cleared, memory can get filled up over time.

Fragmentation:
Continuous creation and deletion of objects can break memory into small pieces, making it harder to reuse efficiently.

Reference cycles:
When two or more objects refer to each other, Python’s garbage collector may not free them easily, causing wasted memory.

High memory usage of objects:
Python objects generally take more memory compared to lower-level languages like C or C++.



```python
#Ques3-->Write a Python program that logs an error message to a log file when a
#division by zero exception occurs.
import logging

# Configure the logging system
logging.basicConfig(filename="error.log", level=logging.ERROR,
                    format="%(asctime)s - %(levelname)s - %(message)s")

try:
    # Code that may cause division by zero
    num1 = int(input("Enter numerator: "))
    num2 = int(input("Enter denominator: "))
    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError as e:
    print("Error: Division by zero is not allowed.")
    logging.error("Division by zero occurred: %s", e)
```

Ques7-->:Explain the difference between apply() and map() in Pandas?

-->Difference between apply() and map() in Pandas

1.Purpose and Scope:

map() is mainly used with a Series to apply a function element-wise. Each element in the Series is transformed individually.

apply() can be used with both Series and DataFrame. For a Series, it works element-wise like map(). For a DataFrame, it can apply a function either row-wise or column-wise.

2.Input Flexibility:

map() only works with functions, dictionaries, or Series for element-wise mapping.

apply() can handle more complex functions that may return scalars, lists, or Series, making it more flexible.

3.DataFrame Usage:

map() cannot be used directly on a DataFrame.

apply() can be used on a DataFrame to apply functions across rows (axis=1) or columns (axis=0).

4.Return Type:

map() always returns a Series.

apply() can return a Series or a DataFrame, depending on the input and function applied.

```python
#Ques10-->Calculate the correlation matrix using Seaborn and visualize it with a
#heatmap.

# Import required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data (or load from CSV)
data = {
    "Math": [90, 80, 70, 85, 95],
    "Physics": [85, 78, 72, 88, 92],
    "Chemistry": [88, 82, 75, 90, 94],
    "Biology": [70, 75, 80, 65, 78]
}

df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()
print("Correlation Matrix:\n", corr_matrix)

# Visualize correlation matrix with heatmap
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```
